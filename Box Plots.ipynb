```python
import numpy as np
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
tips = pd.read_csv('tips.csv')
```

```python
tips.head()
```

```python
sns.boxplot(x=tips['size'])
```

```python
sns.boxplot(x=tips['total_bill'])
```

```python
sns.boxplot(x='sex', y='total_bill', data=tips)
```

```python
sns.boxplot(x='day', y='total_bill', data=tips)
```

```python
sns.boxplot(x='day', y='total_bill', data=tips, hue='sex', palette='husl')
```

```python
sns.boxplot(x='day', y='total_bill', data=tips, hue='smoker', palette='coolwarm')
```

```python
sns.boxplot(x='day', y='total_bill', data=tips, hue='time')
```

```python
sns.boxplot(x='day', y='total_bill', data=tips, order=['Sun', 'Fri', 'Thur', 'Sat'], palette='husl')
```

```python
sns.boxplot(x='sex', y='tip', data=tips)
```

```python
sns.boxplot(x='sex', y='tip', data=tips, order=['Male', 'Female'])
```
