```python
import numpy as np
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
tips = pd.read_csv('tips.csv')
```

```python
sns.barplot(x='day', y='tip', data=tips)
```

```python
sns.barplot(x='day', y='total_bill', data=tips)
```

```python
sns.barplot(x='day', y='tip', data=tips, hue='sex')
```

```python
sns.barplot(x='day', y='tip', data=tips, hue='sex', palette='winter_r')
```

```python
sns.barplot(x='day', y='tip', data=tips, hue='smoker')
```

```python
sns.barplot(x='total_bill', y='day', data=tips)
```

```python
sns.barplot(x='total_bill', y='day', data=tips, hue='sex', legend=False, palette='spring')
```

```python
sns.barplot(x='day', y='tip', data=tips,hue='day',  palette= 'spring', legend=False, order=['Sat','Fri','Sun','Thur'])
```

```python
sns.barplot(x='day', y='total_bill', data=tips,)
```
